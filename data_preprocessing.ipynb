Ici on vas lire les données(d'un fichier de type csv) avec pandas data frame

```python
df = pd.read_csv("dataset/credit_risk.csv")
```

Premièrement en vas supprimer les double

```python
df = df.drop_duplicates(subset=['Age', 'Income', 'Home', 'Emp_length', 'Intent', 'Amount', 'Rate',
       'Status', 'Percent_income', 'Default', 'Cred_length'])
df.shape
```

```python
df['Emp_length'] = df['Emp_length'].fillna(df['Emp_length'].median())
df['Rate'] = df['Rate'].fillna(df['Rate'].median())
df.isnull().sum().sort_values(ascending=False)
```

Après la suppression des doublons, la colonne `Rate` contient 3 095 valeurs manquantes (9,55 %) et `Emp_length` en contient 887 (2,74 %). Pour conserver ces observations, nous remplaçons les valeurs manquantes par la médiane de chaque colonne. La médiane est moins sensible aux valeurs extrêmes que la moyenne ; c’est particulièrement utile pour `Emp_length`, qui contient une valeur de 123 ans à vérifier. Nous évitons de remplacer les valeurs manquantes par 0, car zéro pourrait représenter une ancienneté réelle et n’est pas une valeur pertinente pour un taux d’intérêt.

```python
df = df[df["Age"]<120]
```
